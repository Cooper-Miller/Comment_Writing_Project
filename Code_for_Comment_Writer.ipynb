### Comment Writing Engine

#### an application to help teachers write better comments in less time


```python
# functions and code for reading data files, generating comments and writing them to files
# Testing data created by ChatGPT
import csv as csv

def year_qual(gb, improvement):
    '''
    defines the enire quality of your year via 2 variables
    by connor
    '''
    if gb == "good":
        if improvement == "improved":
            return(1)
        else:
            return(2)
    elif gb == "mid":
        if improvement == "improved":
            return(3)
        else:
            return(4)
    elif gb == "bad":
        if improvement == "improved":
            return(5)
        else:
            return(6)
        
def goodbad(gpa):
    '''
    Determines if your grade in the class is good or not
    by connor
    '''
    if gpa > 3.5:
        return("good")
    elif gpa > 2.9:
        return("mid")
    else:
        return("bad")
    
    
    
def convert(grade):
    '''
    converts from a letter grade to a quantifiable number grade
    by connor
    '''
    grades = {"A+": 4.0, "A": 4.0, "A-": 3.7, "B+": 3.3, "B": 3.0, "B-": 2.7, "C+": 2.3, "C": 2.0, "C-": 1.7, "D+": 1.3, "D": 1.0, "D-": 0.7, "F": 0.0}
    return(grades[grade])



def improve(g1,g2):
    '''
    determines wether the quarter grade is higher then the semester
    by connor
    '''
    if g1 > g2:
        return("deteriorated")
    else:
        return("improved")

    
    
    
def gb_comment(goodbad, name):
    '''
    returns a good comment if the gpa the gpa is determined good by goodbad 
    and bad or middle if it is determined by goodbad
    by Cooper
    '''
    if goodbad == 'good':
        return(f'{name}, you have shown yourself to be an outstanding and supporting member in our class.')
    if goodbad == 'mid':
        return(f'{name}, you have shown that you can meet the expectations in our class and are able to participate.')
    if goodbad == 'bad':
        return(f'{name}, you have demonstarted a bad influence on your classmates and yourself. This reflects poorly on you.')
  

    
def greeting(name, quality):
    '''
    gives a greeting based on your year quality
    by Connor
    
    '''
    if quality == 1:
        return(f"Good day {name}, I hope this report card finds you in good health because of how well you did this year.")
    elif quality == 2:         
        return(f"I hope the day is treating you well {name}. This was a good year for you.")
    elif quality == 3:
        return(f"Hello {name}, you put in good effort this year.")
    else:
        return(f"Hi.")
    
def course(tclass):
    '''
    Finds the class the student is in and returns the description
    By Connor 
    '''
    return(classdic[tclass])
    
    

def write_comment(line):
    '''
    Writes the comment by taking in every line and defining all variables aswell as running th functions
    by Cooper and Connor
    '''
    ln = line[0]
    fn = line[1]
    work = line[4]
    growth = line[5]
    strength = line[6]
    q1 = convert(line[2])
    sem = convert(line[3])
    tclass = line[7]
    coursedesc = course(tclass)
    gb = goodbad(sem)
    impdet = improve(q1, sem)
    qual = year_qual(gb, impdet)
    greet = greeting(fn, qual)
    gbcomment = gb_comment(gb, fn)
    comment = f'{greet} {gbcomment} You have {impdet} in {tclass} since the quarter.\nCourse description: \n{coursedesc}\nStrength:\nYour strength is {strength}\nGrowth:\nThe area you have grown the most in is {growth}\nYour final semester grade is {line[3]} which is equal to a {sem} GPA which is {gb}.'
    return comment
    
    
def commentfile(comment, name):
    '''
    Writes the comment to the student text file
    By Cooper
    '''
    with open(f"{name}.txt", 'w') as stufile:
        stufile.write(comment)
        
        

with open("course.csv") as csvfile:
    """
    Creates a classdic with the classes as the key and the description as the other thing
    By Connor
    """
    file = csv.reader(csvfile)
    header = next(file)
    classdic = {}

    for line in file:
        classdic[line[0]] = line[1]


# Writes the comment and makess a txt file for it By cooper
with open("teacher.csv") as csvfile:
    file = csv.reader(csvfile)
    header = next(file)
    
    for line in file:
        
        comment = write_comment(line)
        commentfile(comment, f'{line[1]}{line[0]}')
       

    
   
```
